```python
#Bitcoin daily return calculations

old_price = 60000
new_price = 57000

daily_return = (new_price - old_price)/old_price 

print(daily_return)
print(f"Bitcoin daily return: {daily_return:.2%}")
```

```python
bitcoin_prices = [60000, 63000, 61000, 64000, 62000]

daily_returns = []

for i in range (1, len(bitcoin_prices)):
    daily_return = (bitcoin_prices[i] - bitcoin_prices[i - 1]) / bitcoin_prices[i - 1]
    daily_returns.append(daily_return)
  print(daily_returns)  
```

```python
import numpy as np 

volatility = np.std(daily_returns, ddof=1)
print(f"Daily Volatility: {volatility:.2%}")

```
